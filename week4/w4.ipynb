### Question 3, 11

```python
def init_array():
    '''Initiates an array w/ increasing values based on user input of # of rows and columns'''

    rows_input = input('# of rows? ')
    cols_input = input('# of columns? ')

    rows = int(rows_input)
    cols = int(cols_input)

    test_array = np.zeros((rows, cols), int)
    array_alt = np.reshape(np.arange(1, rows * cols + 1), (rows, cols))

    for r in range(rows):
        for c in range(cols):
            test_array[r][c] = c + 1 + r * (cols)

    print(f'Array:\n{test_array}')
    print(f'Using arange and reshape: {array_alt}')
    return test_array, rows, cols

init_array()
```

### Question 4

```python
print('Select your dimensions. Must be larger than a 3x3 matrix!\n')
test_array = init_array()[0]

# prints last column
print(f'Last column:\n{test_array[:, -1]}')
# prints last row
print(f'Last row:\n{test_array[-1, :]}')

# prints 3x3 of upper left
print(f'Upper left:\n{test_array[:3, :3]}')

# prints 3x3 of upper right
print(f'Upper right:\n{test_array[:3, -3:]}')

# prints transpose of array
print(f'Transpose:\n{test_array.T}')
```
